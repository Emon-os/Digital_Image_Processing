```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread('cat.jpg')

# Convert BGR (OpenCV) to RGB (Matplotlib)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 5))
plt.imshow(img)
plt.title('Cat Image')
plt.axis("off")
plt.show()
```

```python
# Question 4 — Extract red, green and blue channels
```

```python
blue, green, red = cv2.split(img)
zero = np.zeros_like(blue)

blue_image = cv2.merge([blue, zero, zero])
green_image = cv2.merge([zero, green, zero])
red_image = cv2.merge([zero, zero, red])

results = [img, red_image, green_image, blue_image]
titles = ["Original", "Red Channel", "Green Channel", "Blue Channel"]

plt.figure(figsize=(14, 8))

for i, result in enumerate(results):
    plt.subplot(2, 2, i + 1)
    plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Question 5 — Grayscale, binary, separate planes and combine planes
gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, binary_image = cv2.threshold(
    gray_image, 127, 255, cv2.THRESH_BINARY
)

blue_plane, green_plane, red_plane = cv2.split(img)

# OpenCV combines channels in BGR order
combined_image = cv2.merge([
    blue_plane,
    green_plane,
    red_plane
])

plt.figure(figsize=(15, 8))

display_images = [
    img,
    gray_image,
    binary_image,
    red_plane,
    green_plane,
    blue_plane,
    combined_image
]

titles = [
    "Original",
    "Grayscale",
    "Binary",
    "Red Plane",
    "Green Plane",
    "Blue Plane",
    "Combined Image"
]

for i, result in enumerate(display_images):
    plt.subplot(2, 4, i + 1)

    if len(result.shape) == 3:
        plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
    else:
        plt.imshow(result, cmap="gray")

    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Question 6 — Arithmetic and logical operations
image1 = cv2.imread("image.jpg")
image2 = cv2.imread("image2.jpg")

if image1 is None or image2 is None:
    raise FileNotFoundError("image.jpg or image2.jpg was not found")

image2 = cv2.resize(
    image2,
    (image1.shape[1], image1.shape[0])
)

# Arithmetic operations
addition = cv2.add(image1, image2)
subtraction = cv2.subtract(image1, image2)

# Logical operations
and_image = cv2.bitwise_and(image1, image2)
or_image = cv2.bitwise_or(image1, image2)
not_image = cv2.bitwise_not(image1)

results = [
    image1,
    image2,
    addition,
    subtraction,
    and_image,
    or_image,
    not_image
]

titles = [
    "Image 1",
    "Image 2",
    "Addition",
    "Subtraction",
    "AND",
    "OR",
    "NOT Image 1"
]

plt.figure(figsize=(16, 10))

for i, result in enumerate(results):
    plt.subplot(2, 4, i + 1)
    plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```
