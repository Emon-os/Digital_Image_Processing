

```
**Python program for point transformations**
Image negative, log transformations, and power-law transformations.
```



````python
import cv2
import numpy as np
import matplotlib.pyplot as plt
````

````python
img = cv2.imread("cat.png")

# OpenCV reads as BGR, convert to RGB manually
img_rgb = img[:, :, ::-1]

# Display
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
````

````python
I = 0.299 * img[:, :, 0] + 0.587 * img[:, :, 1] + 0.114 * img[:, :, 2]
````

````python
plt.figure(figsize=(12,6))
plt.subplot(1,2, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2, 2)
plt.imshow(I, cmap="gray")
plt.title("Grayscale")
plt.axis("off")
plt.show
````

````python
I_neg = 255-I

plt.figure(figsize=(12,6))
plt.subplot(1,2, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2, 2)
plt.imshow(I_neg, cmap="gray")
plt.title("Negative")
plt.axis("off")
plt.show()
````

````python
# Log transformation function
c = 0.4  # scaling constant
I_log = c * np.log1p(I)  # log(1 + I)
````

````python
plt.figure(figsize=(12,6))
plt.subplot(1,2, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2, 2)
plt.imshow(I_log, cmap="gray")
plt.title("Log Transformation")
plt.axis("off")
plt.show()
````

````python
c = 0.01
I_gamma = c * np.power(I, 0.5)
````

````python
plt.figure(figsize=(12,6))
plt.subplot(1,2, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2, 2)
plt.imshow(I_gamma, cmap="gray")
plt.title("Gamma Transformation")
plt.axis("off")
plt.show()
````
