```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("cat.png")
```

```python

rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

gray_image = cv2.cvtColor(rgb_image, cv2.COLOR_RGB2GRAY)


# Separate RGB Image into R, G, B Planes

R = rgb_image[:, :, 0]
G = rgb_image[:, :, 1]
B = rgb_image[:, :, 2]

# 4. Combine R, G, B Planes

# Merge the three channels
combined_image = cv2.merge([R, G, B])

# Display Results

plt.figure(figsize=(12, 10))

# Original Color Image
plt.subplot(2, 4, 1)
plt.imshow(rgb_image)
plt.title("Original Color Image")
plt.axis("off")

# Grayscale Image
plt.subplot(2, 4, 2)
plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")


# Red Plane
plt.subplot(2, 4, 3)
plt.imshow(R, cmap="Reds")
plt.title("Red (R) Plane")
plt.axis("off")

# Green Plane
plt.subplot(2, 4, 4)
plt.imshow(G, cmap="Greens")
plt.title("Green (G) Plane")
plt.axis("off")

# Blue Plane
plt.subplot(2, 4, 5)
plt.imshow(B, cmap="Blues")
plt.title("Blue (B) Plane")
plt.axis("off")

# Combined RGB Image
plt.subplot(2, 4, 6)
plt.imshow(combined_image)
plt.title("Combined RGB Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
