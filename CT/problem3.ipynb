```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("cat.png")
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# 3 Bit-Plane Slicing
# Extract all 8 bit planes
bit_planes = []

for bit in range(8):
    # Extract the bit
    plane = (gray >> bit) & 1

    # Convert 0/1 to 0/255 for display
    plane = plane * 255

    bit_planes.append(plane.astype(np.uint8))


# Display Bit Planes
plt.figure(figsize=(12, 8))

for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(bit_planes[i], cmap="gray")
    plt.title(f"Bit Plane {i}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
